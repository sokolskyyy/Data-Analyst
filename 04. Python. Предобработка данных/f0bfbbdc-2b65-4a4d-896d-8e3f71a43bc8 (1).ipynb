## 2.  Проверка ошибок в данных и их предобработка


### 2.1. Названия, или метки, столбцов датафрейма


```python
# Выведем все названия столбцов
print(new_games.columns)
```

```python
# Преобразуем столбцы в snake_case (нижний регистр), заменим пробелы на '_'
new_games.columns = (new_games.columns.str.lower().str.replace(' ', '_'))
print(new_games.columns)
```

### 2.2. Типы данных

```python
# Приведем 'year_of_release' к типу данных int с понижением размерности значений
new_games['year_of_release'] = new_games['year_of_release'].fillna(0)
new_games['year_of_release'] = pd.to_numeric(new_games['year_of_release'], downcast='integer')


new_games['critic_score'] = new_games['critic_score'].astype('Int16')
```

```python
# Приведем следующие к типу данных float с понижением размерности значений:
new_games['eu_sales'] = pd.to_numeric(new_games['eu_sales'], errors='coerce', downcast='float')


new_games['jp_sales'] = pd.to_numeric(new_games['jp_sales'], errors='coerce', downcast='float')


new_games['user_score'] = pd.to_numeric(new_games['user_score'], errors = 'coerce', downcast = 'float')
```

```python
# Снизим размерность:
new_games['na_sales'] = pd.to_numeric(new_games['na_sales'], downcast='float')


new_games['other_sales'] = pd.to_numeric(new_games['other_sales'], downcast='float')
```

### 2.3. Наличие пропусков в данных


```python
# Выводим количество пропущенных строк в датафрейме
new_games.isna().sum()
```

```python
# Подсчитываем доли строк с пропусками
new_games.isna().mean()
```

В данных наблюдаются пропуски в следующих столбцах:
- `name`: в 2 строках (0.01% данных) отсутствует информация о названии игр. Это может затруднить идентификацию, если необходимо провести анализ по конкретным играм.
- `genre`: в 2 строках (0.01% данных). Пропуски в этом столбце могут повлиять на анализ по жанру игры. Возможно, пропуски в этом поле связаны с пропусками в `name`, учитывая общее количество пропущенных строк.
- `eu_sales`: в 6 строках (0.03% данных) отсутствуют данные о продажах копий в европе.
- `jp_sales`: в 4 строках (0.02% данных) отсутствуют данные о продажах копий в европе.
- `critic_score`: в 8714 строках (51.39% данных) отсутствуют данные о оценках критиков. 
- `user_score`: в 9268 строках (54.65% данных) отсутствуют данные по оценкам пользователей.
- `rating`: в 6871 строках (40.52% данных) отсутствуют данные о рейтинге игр ассоциации ESRB. 

Есть прямая связь между `critic_score`, `user_score`, `rating`, почти во всех этих стоблцах есть пропуски у около половины строк данных, возможно предположить, что эти игры являются старыми и не получали никаких оценок во время выхода игры или не все игры успели получить оценки 

```python
# Удалим все пропуски без названия и жанра
new_games = new_games.dropna(subset=['name', 'genre'])
```

```python
# Удаляем пропсуки без значений продаж в европе и японии
new_games['eu_sales'] = new_games.groupby(['platform', 'year_of_release'])['eu_sales'].transform('mean').fillna(0)
new_games['jp_sales'] = new_games.groupby(['platform', 'year_of_release'])['jp_sales'].transform('mean').fillna(0)
```

```python
# Заполняем пропуски индикаторами
new_games['critic_score'] = new_games['critic_score'].fillna(-1)
new_games['user_score'] = new_games['user_score'].fillna('tbd')
new_games['rating'] = new_games['rating'].fillna('Unknown')
```

```python
# Проверяем critic_score
print("Минимальное значение critic_score:", new_games['critic_score'].min())

# Проверяем user_score
print(new_games['user_score'].unique()[:10])

# Проверяем rating
print(new_games['rating'].unique())
```

Коллеги хотят изучить историю продаж игр в начале XXI века, и их интересует период с 2000 по 2013 год включительно. Отберите данные по этому показателю. Сохраните новый срез данных в отдельном датафрейме, например `df_actual`.

## 4. Категоризация данных

```python
# Созадем категории для оценок пользователей
def cat_user():
    new_games['user_score_category'] = pd.cut(pd.to_numeric(new_games['user_score'], errors='coerce'), bins=[0,3,8,10], labels=["низкая оценка", "средняя оценка", "высокая оценка"], right=False).cat.add_categories(['нет оценки']).fillna('нет оценки')
    print(new_games['user_score_category'].value_counts())
cat_user()
```

```python
# Созадем категории для оценок критиков
def cat_critic():
    new_games['critic_score_category'] = pd.cut(pd.to_numeric(new_games['critic_score'], errors='coerce'), bins=[0,30,80,100], labels=["низкая оценка", "средняя оценка", "высокая оценка"], right=False).cat.add_categories(['нет оценки']).fillna('нет оценки')
    print(new_games['critic_score_category'].value_counts())
cat_critic()
```
